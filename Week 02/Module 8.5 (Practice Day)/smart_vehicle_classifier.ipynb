```python
from abc import  ABC , abstractmethod
class Vehicle(ABC):
  def __init__(self, vehicle_type, name, speed):
    self.__vehicle_type = vehicle_type
    self.__name = name
    self.__speed = speed

  def classify(self):
    if int(self.__speed) < 100:
      return "High-speed Vehicle"
    else:
      return "Normal Vahicle"

  def get_type(self):
    return self.__vehicle_type

  def set_type(self, new_type):
    self.__vehicle_type = new_type

  def get_name(self):
    return self.__name

  def set_name(self, new_name):
    self.__name = new_name

  def get_speed(self):
    return self.__speed

  @abstractmethod
  def show_info(self):
    pass
```

```python
class Car(Vehicle):
  def __init__(self, type, name, speed):
    super().__init__(type, name, speed)

  def show_info(self):
    print(f"Type: {self.get_type()} Name: {self.get_name()} Speed: {self.get_speed()} Predicted: {self.classify()}")
```

```python
class Bike(Vehicle):
  def __init__(self, type, name, speed):
      super().__init__(type, name, speed)

  def show_info(self):
    print(f"Type: {self.get_type()} Name: {self.get_name()} Speed: {self.get_speed()} Predicted: {self.classify()}")
```

```python
n = int(input())

vehicles = []

for i in range(n):
  values = input().split()
  if values[0] == "Car":
    v = Car(values[0], values[1], values[2])
  elif values[0] == "Bike":
    v = Bike(values[0], values[1], values[2])
  else:
    continue

  vehicles.append(v)
  print(f"{v.get_type()} Added: {v.get_name()}")

print("--- Vehicle Details ---\n\n")
for v in vehicles:
  v.show_info()
```
