```python
with open("./sample_data/names.txt", "r") as file:

  names = file.readlines()

  strip_name = list(map(str.strip, names))

  filter_name = list(filter(lambda x : len(x) > 0, strip_name))

  cp_names = list(map(str.capitalize, filter_name))

  unique_name = list(set(cp_names))

  sorted_names = sorted(unique_name)

with open("./sample_data/clean_names.txt", "w") as file:
  for name in sorted_names:
    file.write(name + '\n')
```
