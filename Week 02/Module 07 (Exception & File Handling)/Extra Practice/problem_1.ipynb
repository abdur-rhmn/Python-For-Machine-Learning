```python
with open("./sample_data/emails.txt", "r") as file:

  emails = file.readlines()

  format_emails = list(map(str.strip, emails))

  lower_emails = list(map(str.lower, format_emails))

  unique_emails = list(set(lower_emails))

  sorted_emails = sorted(unique_emails)

with open("./sample_data/unique_emails.txt", "w") as file:
  for email in sorted_emails:
    file.write(email + '\n')
```
