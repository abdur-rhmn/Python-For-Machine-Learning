# Practice 1 :

Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce

with open("./sample_data/sample.txt", "r") as file:

  # Number of lines
  lines = file.readlines()
  total_lines = len(lines)
  print(f"Number of lines : {total_lines}")

  # Number of words
  words = list(map(lambda x : len(x.split()), lines))
  total_words = reduce(lambda x,y : x+y, words)
  print(f"Number of words : {total_words}")

  # Number of characters
  string = list(map(str.strip, line))
  string = list(map(lambda x : x.replace(" ", ""), string))
  characters = list(map(lambda x : len(x), string))
  total_characters = reduce(lambda x,y : x+y, characters)
  print(f"Number of characters : {total_characters}")

with open("./sample_data/counter_of_string.txt", "w") as file:
  file.writelines(f"Number of lines : {total_lines}\nNumber of words : {total_words}\nNumber of characters : {total_characters}")
```
