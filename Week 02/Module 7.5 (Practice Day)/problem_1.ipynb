```python
a_friend = input().split()

b_friend = input().split()

a_friend = set(a_friend)

b_friend = set(b_friend)

mutual_friend = a_friend.intersection(b_friend)

unique_a = a_friend.difference(b_friend)

unique_b = b_friend.difference(a_friend)

total_unique_friends = a_friend.union(b_friend)

print(f"Mutual friends: {mutual_friend}")
print(f"Unique to A: {unique_a}")
print(f"Unique to B: {unique_b}")
print(f"Total unique friends: {len(total_unique_friends)}")
```
