```python
class MobilePhone:
  def __init__(self, model, battery = 100):
    self.model = model
    self.__battery = battery

  def use(self, hour):
    if self.__battery > 0:
      self.__battery -= (hour*2)
      if self.__battery <= 0:
        self.__battery = 0
        print("Low battery")
    else:
      print("Low battery")

  def charge(self, hour):
    if self.__battery < 100:
      self.__battery += (hour*5)
      if self.__battery >= 100:
        self.__battery = 100
        print("Fully charged")
    else:
      print("Fully charged")

  def get_battery(self):
    print(self.__battery)
```

```python
nokia = MobilePhone("Nokia 3310")

nokia.get_battery()

nokia.use(51)

nokia.get_battery()

nokia.charge(7)

nokia.get_battery()
```
