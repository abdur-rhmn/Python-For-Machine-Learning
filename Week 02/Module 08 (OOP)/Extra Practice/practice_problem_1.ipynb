```python
class phone:
  def __init__(self, model, battery = 100):
    self.model = model
    self.battery = battery
```

```python
class SmartPhone(phone):
  def __init__(self, model, os):
    super().__init__(model)
    self.os = os
```

```python
class GamingPhone(SmartPhone):
  def __init__(self, model, os, cooling_system):
    super().__init__(model, os)
    self.cooling_system = cooling_system

  def start_game(self, name):
    print(f"Playing {name} on {self.model}")
```

```python
iphone = GamingPhone("Iphone 17", "IOS", "Vapor Chamber")

iphone.start_game("clash of clan")

samsung = SmartPhone("S24 Ultra", "Android")

print(samsung.model)

print(iphone.battery)
print(samsung.battery)
```
