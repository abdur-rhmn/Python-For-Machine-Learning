```python
def multiple_value(*args):
    print(type(args))
    for val in args:
        print(val)

multiple_value(10,20,30,40,50)
```

```python
def key_value(**kwargs):
    print(type(kwargs))
    for key,val in kwargs.items():
        print(f"{key} : {val}")

key_value(name = "abdur rahman", age = 30)    # pass arguments by variable = value, then it's make them a dictionary.
```
