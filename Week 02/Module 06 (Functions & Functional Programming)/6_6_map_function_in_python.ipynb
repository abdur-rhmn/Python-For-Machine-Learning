```python
lst = [1,2,3,4,5,6]

new_lst = list(map(lambda x : x**2, lst))

print(new_lst)
```

```python
lst = [1,2,3,4,5,6]

new_lst = list(map(lambda x : x**2 if x%2 == 0 else x, lst))

print(new_lst)
```
