```python
from functools import reduce
```

```python
lst = [1,2,3,4,5,6]

summation = reduce(lambda x,y : x+y, lst)

print(summation)
```

```python
lst = [1,2,3,4,10,5,6]

max_value = reduce(lambda x,y : x if x>y else y, lst)

print(max_value)
```

```python
lst = ["Hello,", "Welcome", "to", "programmer", "world"]

string = reduce(lambda x,y : x+" "+y, lst)

print(string)
```
