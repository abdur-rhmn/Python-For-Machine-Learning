```python
numbers = [x for x in range(100)]

filter_even = list(filter(lambda x : x%2==0, numbers))

print(filter_even)

fifty_upper = list(filter(lambda x : x > 50, numbers))

print(fifty_upper)
```

```python
data = [0, 1, '',None, 'Hello', [], [1, 2], None, True, False, None]

cleaned_data = list(filter(None, data))

print(cleaned_data)
```

```python
string = "Welcome to the age of AI"

string = string.lower()

vowel = list(filter(lambda x : x in "aeiou", string))

print(vowel)
```

```python
fruits = [ 'aaa' , 'bb' , 'cccc' , 'ddddd' ]

filtering = list(filter(lambda x: len(x)>2 , fruits))

print(filtering) 
```
