```python
def num():
    return 10

type(num())
```

```python
def num():
    return 10,20,30

print(type(num()))

# unpacking

a,b,c = num()
print(a,b,c)
```

```python
def give_prediction():

    a= [10,20,30]
    b= [20,30,40]
    return a,b 

type(give_prediction())

## unpacking 
x , y = give_prediction()

print(x, y)
```

```python
def give_prediction():

    a= {10,20,30}
    b = [100,200,300]
    return a,b 

x = give_prediction()

print(type(x))

for item in x:
    print(item) 
```

```python
def give_prediction():
    return { "name" : "adil" , "addr" : "Ctg"},{ "name" : "abrar" , "addr" : "dhaka"}


x,y = give_prediction()

print(type(x))


for item in x.items():
    print(item)

for item in y.items():
    print(item)
```
