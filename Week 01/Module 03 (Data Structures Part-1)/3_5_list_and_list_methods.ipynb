```python
int_numbers = [10,20,30,40,50]
float_numbers = [3.45, 5.35, 8.65]
str_list = ["apple", "banana", "orange"]

mixed_list = [10, 4.53, "apple"]

two_D_list = [[10,20,30], [40,50,60], [70,80,90]]

print(int_numbers)
print(float_numbers)
print(str_list)

print(mixed_list)

print(two_D_list)
print(two_D_list[1])
print(two_D_list[1][2])
```

```python
## access

numbers = [10,20,30,40,50]

print(numbers[2])

numbers[2] = 300

print(numbers[2])

new_list = numbers[2:5]

print(new_list)
```

```python
## modification

## add items

numbers = [10,20,30,40,50,100,100]

numbers.append(60)

print(numbers)

numbers.insert(1,100)

print(numbers)

## deleting

numbers.pop()

print(numbers)

numbers.remove(100)

print(numbers)
```

```python
## shorting

numbers = [50,30,20,400,60]

numbers.sort()

print(numbers)

numbers.sort(reverse=True)

print(numbers)

numbers.reverse()

print(numbers)
```

## List as Stack

```python
stack = []

# insertion
stack.append(1)
stack.append(2)
stack.append(3)

print(f"TOP value: {stack[-1]}")

stack.pop()

print(f"TOP value: {stack[-1]}")

stack.pop()

print(f"TOP value: {stack[-1]}")
```

## List as Queue

```python
queue = []

# insertion
queue.append(1)
queue.append(2)
queue.append(3)

print(f"Front value: {queue[0]}")

queue.pop(0)

print(f"Front value: {queue[0]}")

queue.pop(0)

print(f"Front value: {queue[0]}")
```

## List Comprehension

```python
even = [x for x in range(11) if x%2==0]

random_list = [x*2 for x in range(5,11)]

print(even)

print (random_list)
```
