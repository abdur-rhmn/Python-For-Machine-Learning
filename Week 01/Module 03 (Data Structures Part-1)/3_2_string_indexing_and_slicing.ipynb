## String methods

```python
string = "HOw Are yOu?, How do you feel?"

string_capitalize = string.capitalize()
print(string_capitalize)

string_upper_case = string.upper()
print(string_upper_case)

string_lower_case = string.lower()
print(string_lower_case)

# lenght
print("Lenght:", len(string))

# find a substring
print("are" in string_lower_case)
print("the" in string_lower_case)

index = string_lower_case.index("are")
print(index)

count = string_lower_case.count("how")
print(count)

shuru_index = string_lower_case.find("you")
print(shuru_index)

sesh_index = string_lower_case.rfind("you")
print(sesh_index)

# replace
new_string = string_lower_case.replace("feel", "doing")
print(new_string)

```

### Spliting, joining and Formating method

```python
prompt = "What is your name?"

tokens = prompt.split()
print(tokens)

sentance = " ".join(tokens)
print(sentance)
```
