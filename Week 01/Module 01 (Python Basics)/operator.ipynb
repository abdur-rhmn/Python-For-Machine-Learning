# Arethmetic Operator

```python
x = 10; y = 3

sum = x + y
sub = x - y
div0 = x/y
div1 = round(x/y)
div2 = round(x/y, 2)
mul = x*y

mod = x%y

print(sum, sub, div0, mul, mod)
print(div1, div2)
print(f"{div0:.2f}")
```
