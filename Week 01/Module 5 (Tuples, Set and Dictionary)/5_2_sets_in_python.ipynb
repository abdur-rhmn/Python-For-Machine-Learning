```python
S = {3,1,2,4,5,1,2,3}

if 3 in S:
    print("3 in set")
else:
    print("not present")

for element in S:
    print(element, end = " ")
```

```python
s = {3,1,2,4,5,1,2,3}

s.add(10)
s.add(0)

print(s)

s.pop()
print(s)
s.pop()
print(s)

s.remove(5)
print(s)
```

```python
a = {1,2,3}
b = {1,2,3,4}
c = {5,6,7}

print(a.union(b))
print(a.intersection(b))
print(a.isdisjoint(b))
print(a.isdisjoint(c))
print(a.issubset(b))
print(b.issubset(c))
```
