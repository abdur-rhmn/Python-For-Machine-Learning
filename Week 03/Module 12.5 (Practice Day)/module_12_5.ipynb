## problem 7

```python
data = {
    "Name": [
        "Alice", "Bob", "Charlie", "Diana",
        "Evan", "Fiona", "George", "Hannah"
    ],
    "Score": [88, 74, 77, 55, 85, 49, 33, 72]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.loc[(df['Score'] > 60)]
```

```python
df.loc[(df['Score'] < 40)]
```

```python
df.loc[(df['Score'] > 60) | (df['Score'] < 40)]
```
