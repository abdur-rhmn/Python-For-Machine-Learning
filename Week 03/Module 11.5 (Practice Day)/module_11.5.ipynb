## problem_1

```python
arr1 = np.array([1,2,3])

arr2 = np.array([[1,2,3],[4,5,6]])

arr3 = np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])

print(arr1.ndim)
print(arr1.shape)
print(arr1.size)
print(arr1.dtype)

print(arr2.ndim)
print(arr2.shape)
print(arr2.size)
print(arr2.dtype)

print(arr3.ndim)
print(arr3.shape)
print(arr3.size)
print(arr3.dtype)
```

## problem_2

```python
z = np.zeros((3,3))
print(z)
print(z.shape)

o = np.ones((2,5))
print(o)
print(o.shape)

# 1D
f = np.full((5,),7)
print(f)
print(f.shape)

# 2D
f = np.full((3,4),7)
print(f)
print(f.shape)

# 3D
f = np.full((3,3,4),7)
print(f)
print(f.shape)
```
