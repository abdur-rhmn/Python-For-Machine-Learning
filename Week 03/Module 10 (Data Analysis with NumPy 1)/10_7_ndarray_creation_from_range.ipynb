```python
import numpy as np
```

```python
# np.arange(first , end , step ) -> similar to range function 

arr = np.arange(0, 10, 1).reshape(2,5)

# mat = arr.reshape(2,5)

print(arr)
```

```python
# np.linspace(start , end(included) , koita) 

arr = np.linspace(1, 10, 20)    # Picked 20 ta linear values between 1 and 10.

print(arr)
print(arr.dtype)
```

```python
# np.logspace(start , end(included) , koita )

arr = np.logspace(1, 10, 5)    # base 10

arr2 = np.logspace(0, 4, 6, base = 2)

print(arr)

print(arr2)
```
