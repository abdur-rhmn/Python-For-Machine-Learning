```python
import numpy as np

arr1 = np.array([1,2,3,4,5])
```

```python
# veiw the array not copy
v_arr = arr1[1:4]

v_arr[1] = 30

print(arr1)
print(v_arr)
```

```python
# copy
arr1 = np.array([1,2,3,4,5])
cpy_arr = arr1[1:4].copy()

cpy_arr[1] = 30

print(arr1)
print(cpy_arr)
```

## Advance Indexing

```python
lst = np.array([10,20,30,40,50])

values = lst[[0,3,1]]

print(values)
```

```python
lst = np.array([[10,20,30],[40,50,60]])

# name_of_list[[row], [column]]

values = lst[[0,1], [1,2]]

print(values)
```

```python
# boolean way
print(lst[lst>20])
```

```python
lst[lst <= 20] = 0

print(lst)
```

## Iteration

```python
arr = np.array([[10,20,30],[40,50,60]])

for i in np.nditer(arr):
    print(i)
```
