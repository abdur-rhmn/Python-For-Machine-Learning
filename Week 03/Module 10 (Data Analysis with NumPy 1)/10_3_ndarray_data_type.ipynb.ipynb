```python
import numpy as np
arr = np.array([1,2,3]) 

print(arr.dtype)

# upcasted to float 
arr = np.array([1.5,2.4,3.2]) 
print(arr.dtype)
```

```python
#upcasted to string
arr = np.array([1,2,3.2,'hello']) 
print(arr.dtype) 


arr = np.array([1,True,3.25,'hello',]) 
print(arr.dtype) 
```

```python
arr = np.array([1,2,300],dtype=np.int16)
print(arr.dtype)
print(arr)




arr = arr.astype(np.int32)
print(arr.dtype)


#error
arr = np.array([1,2,3.2,'hello'],dtype=np.float64) 
```
