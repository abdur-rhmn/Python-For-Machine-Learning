```python
# You must run this code and use that data set for all the questions
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

# **1) NumPy — 5 Questions**

## Ans to the Q1

```python
ages = np.array(ages)
salaries = np.array(salaries)

print(ages)
print(salaries)
```

```python
print(ages.dtype)
print(salaries.dtype)

print(ages.ndim)
print(salaries.ndim)

print(ages.shape)
print(salaries.shape)

print(ages.size)
print(salaries.size)
```

## Ans to the Q2

```python
highest_salary = np.max(salaries)
lowest_salary = np.min(salaries)

print(f"Highest salary : {highest_salary}")
print(f"Lowest salary : {lowest_salary}")
```

```python
average_salary = np.mean(salaries)
average_age = np.mean(ages)

print(f"Average salary : {average_salary}")
print(f"Average age : {average_age}")
```

## Ans to the Q3

```python
above_30 = ages[ages > 30]

print(above_30)
print(f"{above_30.size} employees are older than 30")
```

## Ans to the Q4

```python
updated_ages = ages+5

print(updated_ages)
```

## Ans to the Q5

```python
total_salary_expense = np.sum(salaries)

print(total_salary_expense)
```

```python
diff = np.max(salaries) - np.min(salaries)
print(diff)
```
