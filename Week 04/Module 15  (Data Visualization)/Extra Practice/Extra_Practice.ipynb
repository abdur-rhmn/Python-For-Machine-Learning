```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('student_IQdata.csv')
df
```

```python
plt.scatter(df['Study_Hour'], df['IQ_Score'])
plt.title("Study Hours vs IQ Score")
plt.xlabel("Study Hours")
plt.ylabel("IQ Score")
plt.show()
```

```python
plt.scatter(df['Shoe_Size'], df['Study_Hour'])
plt.title("Shoe Size vs Study Hours")
plt.xlabel("Shoe Size")
plt.ylabel("Study Hours")
plt.show()
```

```python
plt.hist(df['Shoe_Size'], edgecolor='black')
plt.title("Distribution of Shoe Size")
plt.xlabel("Shoe Size")
plt.ylabel("Frequency")
plt.show()
```
