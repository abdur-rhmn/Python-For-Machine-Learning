```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Problem 1

```python
df = pd.read_csv('Practice Day.csv')

df
```

```python
df['Discounted_Price'] = df['Price'] - ((df['Price'] * 10) / 100)

df.to_csv('Discounted_Price.csv')
```

# problem 2

```python
print(df['EmployeeID'].duplicated().count())
```

```python
df.drop_duplicates(subset=['EmployeeID'], inplace=True)

df
```

# Problem 3

```python
print(df[['Age', 'Salary']].isnull().sum())
```

```python
df.fillna({'Age' : df['Age'].mean().round()}, inplace=True)

df
```

```python
df.dropna(subset=['Salary'], inplace=True)
df
```

# Problem 4

```python
Marks1_mean = df['Marks1'].mean().round()
Marks1_max = df['Marks1'].max()

print(Marks1_mean, Marks1_max)

Marks2_mean = df['Marks2'].mean().round()
Marks2_max = df['Marks2'].max()

print(Marks2_mean, Marks2_max)

Marks3_mean = df['Marks3'].mean().round()
Marks3_max = df['Marks3'].max()

print(Marks3_mean, Marks3_max)
```

```python
df['Total Marks'] = df[['Marks1', 'Marks2', 'Marks3']].sum(axis=1)

df
```

```python
df['Total Marks'].std()
```
